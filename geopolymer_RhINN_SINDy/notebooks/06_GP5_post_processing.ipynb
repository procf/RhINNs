# GP5 — post-processing of the identified constitutive equations

Integrates eq. (7) with the coefficients of Tables 3 and 5 (`results/coefficients/`) and compares
it with the experiments. Panels are chosen by sample age, as labelled in the tables.

| Output | Paper |
|---|---|
| `figures/fig04_GP5_SAOS_lissajous.png` | Fig. 4 |
| `figures/fig06_GP5_LAOS_lissajous.png` | Fig. 6 |
| `figures/fig08_GP5_startup.png` | Fig. 8 |
| `figures/figS_GP5_coefficients.png` | coefficients vs. their physical analogues (Table 2) |

```python
import matplotlib.pyplot as plt

from rhinn_sindy import figures
from rhinn_sindy.coefficients import load_laos_coefficients
from rhinn_sindy.plotting import save_figure, set_paper_style

MATERIAL = "GP5"
set_paper_style()
load_laos_coefficients(MATERIAL)
```

## Fig. 4 — SAOS Lissajous curves (0.02 % strain, 1 rad/s)

The measured loop is rebuilt from the stress amplitude and phase angle of the time sweep;
the model loop is the steady cycle of $\dot\gamma = c_1\sigma + c_2\dot\sigma$.

```python
# Ages in minutes (Table 3).
fig = figures.saos_lissajous(MATERIAL, ages=[3, 7.5, 27, 37], nrows=1, ncols=4)
save_figure(fig, "fig04_GP5_SAOS_lissajous.png", pad_inches=0.15)
plt.show()
```

## Fig. 6 — LAOS Lissajous curves (100 Pa, 1 rad/s)

The fitted imposed stress drives eq. (7), integrated from the first measured strain.

```python
# Ages in minutes (Table 5).
fig = figures.laos_lissajous(MATERIAL, ages=[6, 32, 50, 76], nrows=1, ncols=4)
save_figure(fig, "fig06_GP5_LAOS_lissajous.png", pad_inches=0.15)
plt.show()
```

## Fig. 8 — start-up shear at 24, 50 and 76 min

A protocol not used in training: constant shear rate from rest. It checks that the
identified equations stay bounded.

```python
fig = figures.startup(MATERIAL, ages=[24, 50, 76], ylabel_all=True)
save_figure(fig, "fig08_GP5_startup.png")
plt.show()
```

## Coefficients vs. their physical analogues (Table 2)

Each coefficient is plotted with the combination of η and G′ it represents.

```python
fig = figures.coefficient_analogues(MATERIAL)
save_figure(fig, "figS_GP5_coefficients.png")
plt.show()
```
