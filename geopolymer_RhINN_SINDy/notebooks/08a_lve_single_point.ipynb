# Single-point linear-regime calibration of GP2–GP4 (Section 3)

GP2–GP4 have no SAOS time sweep. For each, one row on the linear plateau of the amplitude
sweep (10 rad/s) is turned into stress and strain waveforms, and
$\dot\gamma = c_1\sigma + c_2\dot\sigma$ is identified with the RhINN + STRidge solvers.
These $(c_1, c_2)$ scale the GP1 age profile in `08c_3ITT_GP2-GP4`. The rows and training
settings are in `rhinn_sindy/settings.py`.

The sweep stores only amplitudes and the phase angle $\delta$, so the waveforms are rebuilt as
$\sigma_0\sin\omega t$ and $\gamma_0\sin(\omega t \pm \delta)$. The paper used $+\delta$
(strain leading the stress), which gives $c_1 < 0$. For a Maxwell model the exact answer is

$$c_1 = \mp\frac{\gamma_0\omega}{\sigma_0}\sin\delta,\qquad c_2 = \frac{\gamma_0}{\sigma_0}\cos\delta ,$$

which is written next to the trained values as a check.

* **Output:** `results/retrained/lve_single_point.csv`
* **Runtime:** roughly 10–20 min per composition on a laptop CPU.
* **Hyperparameters:** the settings in `rhinn_sindy/settings.py` are the ones used for the
  paper. Another amplitude-sweep row or composition has different stress and strain
  magnitudes and may need some hyperparameter tuning, mainly the loss weights, learning
  rate, number of rounds and STRidge tolerance.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf

from rhinn_sindy import figures
from rhinn_sindy.paths import RETRAINED_DIR
from rhinn_sindy.reproduce import run_lve

tf.random.set_seed(1234)
np.random.seed(1234)
```

```python
# +1 reproduces the paper (strain leads stress); -1 is the physically lagging strain (c1 > 0).
PHASE_SIGN = +1
```

## Train

```python
results = run_lve(["GP2", "GP3", "GP4"], phase_sign=PHASE_SIGN)
pd.read_csv(RETRAINED_DIR / "lve_single_point.csv")
```
